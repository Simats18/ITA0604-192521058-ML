```python

import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture

# 1. Create sample dataset
np.random.seed(42)

X1 = np.random.normal(5, 1, (100, 1))
X2 = np.random.normal(10, 1, (100, 1))
X = np.vstack((X1, X2))

# 2. Implement EM using Gaussian Mixture Model
model = GaussianMixture(
    n_components=2,
    random_state=42
)

# 3. Train the model
model.fit(X)

# 4. Predict clusters
labels = model.predict(X)

# 5. Display results
print("Cluster means:")
print(model.means_)

print("\nCluster variances:")
print(model.covariances_)

print("\nFirst 10 cluster labels:")
print(labels[:10])

print("\nLog-likelihood:", model.score(X) * len(X))

# 6. Visualize clusters
plt.scatter(X[:, 0], np.zeros(len(X)),
            c=labels, cmap="viridis")
plt.xlabel("Data values")
plt.title("Expectation-Maximization Clustering")
plt.yticks([])
plt.show()
```
